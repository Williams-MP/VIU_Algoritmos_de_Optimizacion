````python
# Excelente. Aquí tienes el fragmento de código en **Python** que implementa la lógica de filtrado del **Capítulo I** del Reglamento de IA de la UE, diseñado para determinar si tu sistema entra en el ámbito de aplicación o si está exento.

# He incluido los números de **Artículo** en los comentarios para asegurar la trazabilidad legal del código.

# ```python
def evaluar_ambito_aplicacion():
    """
    Evalúa si un sistema está sujeto al Reglamento de IA de la UE 
    basándose en el Capítulo I (Disposiciones Generales).
    """
    print("--- Evaluación de Ámbito de Aplicación (Capítulo I) ---")

    # 1. Definición de Sistema de IA - Artículo 3, punto 1
    # Un sistema basado en una máquina con autonomía que infiere salidas.
    es_sistema_ia = input("¿Es un sistema basado en una máquina con autonomía que genera predicciones o decisiones? (s/n): ").lower() == 's'
    
    if not es_sistema_ia:
        return "FUERA DE ÁMBITO: El sistema no cumple la definición de IA del Art. 3(1)."

    # 2. Vínculo Territorial con la UE - Artículo 2, apartado 1
    # Aplica si se comercializa en la UE o si el uso ocurre en la Unión.
    vinculo_ue = input("¿El sistema se introduce en el mercado de la UE o sus resultados se usan en la UE? (s/n): ").lower() == 's'
    
    if not vinculo_ue:
        return "FUERA DE ÁMBITO: No existe vínculo territorial según el Art. 2(1)."

    # --- FILTROS DE EXCLUSIÓN CRÍTICA ---

    # 3. Exclusión Militar y Seguridad Nacional - Artículo 2, apartado 3
    # Quedan excluidos independientemente de la entidad.
    militar_defensa = input("¿Se utiliza exclusivamente para fines militares, de defensa o de seguridad nacional? (s/n): ").lower() == 's'
    if militar_defensa:
        return "EXENTO: Excluido por fines de defensa o seguridad nacional (Art. 2(3))."

    # 4. Investigación y Desarrollo Científico - Artículo 2, apartado 6
    # Sistemas creados solo para I+D científica están fuera.
    id_cientifica = input("¿Es un sistema desarrollado exclusivamente para investigación científica? (s/n): ").lower() == 's'
    if id_cientifica:
        return "EXENTO: Excluido por fines de investigación científica (Art. 2(6))."

    # 5. Actividades Pre-comerciales - Artículo 2, apartado 8
    # Excluye desarrollo previo a la puesta en servicio, salvo pruebas en condiciones reales.
    fase_precomercial = input("¿Está en fase de desarrollo o prueba antes de comercializarse (sin ser prueba real)? (s/n): ").lower() == 's'
    if fase_precomercial:
        return "EXENTO: Actividades de investigación y desarrollo pre-comercial (Art. 2(8))."

    # 6. Uso Personal no Profesional - Artículo 2, apartado 10
    # Uso privado por personas físicas en actividades no profesionales.
    uso_personal = input("¿Es para uso puramente personal y no profesional? (s/n): ").lower() == 's'
    if uso_personal:
        return "EXENTO: Uso personal no profesional (Art. 2(10))."

    # 7. Regla de Código Abierto - Artículo 2, apartado 12
    # El código abierto está exento a menos que sea de alto riesgo o prohibido.
    codigo_abierto = input("¿Es un sistema de IA de código abierto? (s/n): ").lower() == 's'
    if codigo_abierto:
        print("AVISO: Sistema de código abierto. Solo sujeto a obligaciones si es de alto riesgo o prohibido (Art. 2(12)).")

    # --- OBLIGACIÓN TRANSVERSAL ---
    # Alfabetización en materia de IA - Artículo 4
    print("OBLIGACIÓN DETECTADA: Debe garantizar un nivel suficiente de alfabetización en IA para su personal (Art. 4).")

    return "CONTINUAR: El sistema está dentro del ámbito. Proceder al Capítulo II (Prácticas Prohibidas)."

# Ejecutar el flujo
resultado = evaluar_ambito_aplicacion()
print(f"\nResultado de la evaluación: {resultado}")
# ```

# ### Resumen de la lógica aplicada:
# *   **Filtros de Entrada:** Si el software no cumple con la definición técnica de "IA" o no tiene nexo con el mercado o usuarios de la Unión, el programa termina inmediatamente como "Fuera de Ámbito".
# *   **Ramificaciones de Exclusión:** Se evalúan las excepciones específicas para militares, investigadores, desarrolladores pre-comerciales y usuarios privados. Si alguna es afirmativa, el sistema queda exento de las obligaciones de este Reglamento.
# *   **Caso Especial (Código Abierto):** Si es *Open Source*, el árbol no se detiene, pero se activa una bandera de aviso, ya que la exención no aplica si el sistema es de alto riesgo o realiza prácticas prohibidas.
# *   **Output Técnico:** Si el sistema supera todos los filtros, se registra la necesidad de cumplir con el **Artículo 4** sobre **alfabetización en materia de IA** para el personal responsable.

# ¿Deseas que pasemos ahora al **Capítulo II (Prácticas prohibidas de IA)** para identificar los siguientes filtros de riesgo inaceptable?
````

````python
# Para continuar con la construcción de tu sistema experto, he analizado detalladamente el **Artículo 5 (Prácticas de inteligencia artificial prohibidas)**. Este artículo es crítico porque define los límites de lo que es legalmente aceptable en la Unión Europea; si el sistema cae en una de estas categorías sin cumplir una excepción, el flujo termina en una prohibición absoluta.

# Aquí tienes el análisis técnico-legal y la propuesta de código:

# ### 1. Filtros Booleanos y Excepciones Críticas

# Basándome en el texto del **Artículo 5**, he extraído los siguientes puntos de decisión:

# *   **Manipulación Subliminal:** ¿Utiliza técnicas subliminales o deliberadamente engañosas para alterar sustancialmente el comportamiento y causar perjuicio?.
# *   **Explotación de Vulnerabilidades:** ¿Explota vulnerabilidades debidas a la edad, discapacidad o situación socioeconómica para alterar el comportamiento de forma perjudicial?.
# *   **Puntuación Social (Social Scoring):** ¿Evalúa o clasifica a personas físicas en función de su comportamiento social o personalidad, provocando tratos desfavorables en contextos no relacionados?.
# *   **Evaluación de Riesgo Delictivo:** ¿Evalúa el riesgo de que una persona cometa un delito basándose solo en el perfilado o características de su personalidad?.
#     *   *Excepción:* No está prohibido si apoya una valoración humana basada en hechos objetivos y verificables directamente relacionados con una actividad delictiva.
# *   **Extracción Facial No Selectiva:** ¿Crea o amplía bases de datos faciales mediante la extracción masiva de imágenes de internet o CCTV?.
# *   **Reconocimiento de Emociones en el Trabajo o Educación:** ¿Se usa para inferir emociones de personas en entornos laborales o centros educativos?.
#     *   *Excepción:* Permitido si es por motivos médicos o de seguridad.
# *   **Categorización Biométrica Sensible:** ¿Clasifica personas para deducir raza, opiniones políticas, convicciones religiosas, vida sexual u orientación sexual?.
#     *   *Excepción:* Permitido el etiquetado/filtrado lícito de conjuntos de datos en el ámbito de la garantía del cumplimiento del Derecho.
# *   **Identificación Biométrica en Tiempo Real (RBI):** ¿Se usa identificación biométrica remota "en tiempo real" en espacios públicos para fines policiales?.
#     *   *Excepción:* Solo para búsqueda selectiva de víctimas/desaparecidos, prevención de amenazas terroristas o localización de sospechosos de delitos graves (penas > 4 años).

# ### 2. Variables y Lógica en Python

# Este bloque de código integra las preguntas y las excepciones documentadas en el **Artículo 5**.

# ```python
def analizar_articulo_5():
    """
    Evalúa si un sistema de IA incurre en prácticas prohibidas según el Artículo 5.
    """
    print("\n--- Evaluación de Prácticas Prohibidas (Artículo 5) ---")

    # Art. 5(1)(a): Técnicas subliminales o manipuladoras
    if input("¿El sistema usa técnicas subliminales o engañosas para causar perjuicio sustancial? (s/n): ").lower() == 's':
        return "PROHIBIDO: Manipulación subliminal (Art. 5(1)(a))."

    # Art. 5(1)(b): Explotación de vulnerabilidades
    if input("¿Explota vulnerabilidades (edad, discapacidad, economía) para alterar el comportamiento? (s/n): ").lower() == 's':
        return "PROHIBIDO: Explotación de vulnerabilidades (Art. 5(1)(b))."

    # Art. 5(1)(c): Puntuación ciudadana (Social Scoring)
    if input("¿El sistema realiza puntuación ciudadana basada en comportamiento social? (s/n): ").lower() == 's':
        return "PROHIBIDO: Puntuación social (Art. 5(1)(c))."

    # Art. 5(1)(d): Evaluación de riesgo delictivo (Predictive Policing)
    if input("¿Predice el riesgo de delincuencia basándose únicamente en perfiles o personalidad? (s/n): ").lower() == 's':
        # Sub-pregunta para excepción Art. 5(1)(d)
        excepcion_policial = input("¿Se basa en hechos objetivos y verificables con apoyo de valoración humana? (s/n): ").lower() == 's'
        if not excepcion_policial:
            return "PROHIBIDO: Evaluación de riesgo delictivo basada en perfilado (Art. 5(1)(d))."

    # Art. 5(1)(e): Scraping facial no selectivo
    if input("¿Recopila imágenes faciales de internet o CCTV de forma no selectiva? (s/n): ").lower() == 's':
        return "PROHIBIDO: Extracción facial no selectiva (Art. 5(1)(e))."

    # Art. 5(1)(f): Reconocimiento de emociones
    if input("¿Deduce emociones en el lugar de trabajo o instituciones educativas? (s/n): ").lower() == 's':
        # Sub-pregunta para excepción Art. 5(1)(f)
        uso_medico_seguridad = input("¿El uso es estrictamente médico o de seguridad? (s/n): ").lower() == 's'
        if not uso_medico_seguridad:
            return "PROHIBIDO: Reconocimiento de emociones en trabajo/educación (Art. 5(1)(f))."

    # Art. 5(1)(g): Categorización biométrica sensible
    if input("¿Categoriza personas según raza, religión, política o sexualidad usando datos biométricos? (s/n): ").lower() == 's':
        # Sub-pregunta para excepción Art. 5(1)(g)
        filtrado_policial_licito = input("¿Es etiquetado/filtrado lícito de datos para cumplimiento del derecho? (s/n): ").lower() == 's'
        if not filtrado_policial_licito:
            return "PROHIBIDO: Categorización biométrica sensible (Art. 5(1)(g))."

    # Art. 5(1)(h): Identificación biométrica remota en tiempo real (RBI)
    if input("¿Realiza identificación biométrica remota 'en tiempo real' en espacios públicos con fines policiales? (s/n): ").lower() == 's':
        # Evaluación de excepciones críticas Art. 5(1)(h)(i, ii, iii)
        fines_especificos = input("¿Se limita a buscar víctimas, prevenir terrorismo o localizar autores de delitos graves (>4 años de pena)? (s/n): ").lower() == 's'
        if fines_especificos:
            # Art. 5(3): Requisito de autorización judicial
            autorizacion = input("¿Cuenta con autorización judicial previa o administrativa independiente vinculante? (s/n): ").lower() == 's'
            if not autorizacion:
                return "PROHIBIDO: RBI sin autorización judicial previa (Art. 5(3))."
            print("AVISO: El uso de RBI está sujeto a cumplimiento estricto de salvaguardias (Art. 5(2)).")
        else:
            return "PROHIBIDO: Identificación biométrica en tiempo real sin fin justificado (Art. 5(1)(h))."

    return "CONTINUAR: El sistema no incurre en prácticas prohibidas. Proceder al Capítulo III (Alto Riesgo)."
# ```

# ### 3. Destino del Flujo

# *   **Si el sistema coincide con alguna prohibición:** El flujo termina inmediatamente con un resultado de **RIESGO INACEPTABLE**. El sistema no puede introducirse en el mercado ni ponerse en servicio en la UE.
# *   **Si el sistema supera todas las comprobaciones (devuelve "CONTINUAR"):** El sistema debe ser evaluado ahora bajo las reglas de clasificación de **Alto Riesgo (Capítulo III, Artículo 6)**.

# ¿Deseas que guardemos este bloque y procedamos con el **Artículo 6** para clasificar si tu sistema es de **Alto Riesgo**?

resultado = analizar_articulo_5()
print(f"\nResultado de la evaluación: {resultado}")
````

````python
# Perfecto. Procedemos ahora con el análisis técnico-legal del **Artículo 6 (Reglas de clasificación de los sistemas de IA de alto riesgo)**. Este artículo es la piedra angular para determinar si un sistema debe cumplir con las obligaciones técnicas más rigurosas de la ley.

# Aquí tienes el desglose para tu arquitectura de software:

# ### 1. Filtros Booleanos (Sí / No)

# Basándonos en el **Artículo 6**, un sistema se clasifica como de alto riesgo por dos vías principales:

# **Vía A: IA como Producto o Componente de Seguridad (Art. 6, apartado 1)**
# *   **¿Es un producto o componente de seguridad?**: ¿El sistema de IA está destinado a ser utilizado como componente de seguridad de un producto regulado en el **Anexo I**, o es el sistema en sí mismo dicho producto?.
# *   **¿Evaluación de terceros?**: ¿Dicho producto está sujeto a una evaluación de conformidad por parte de un **tercero** (organismo notificado) según la legislación de armonización de la Unión del Anexo I?.
#     *   *Nota: Ambas deben ser "Sí" para que sea alto riesgo por esta vía.*

# **Vía B: Casos de uso específicos (Art. 6, apartado 2)**
# *   **¿Se incluye en el Anexo III?**: ¿El sistema se utiliza en ámbitos como biometría, infraestructuras críticas, educación, empleo, acceso a servicios esenciales, garantía del cumplimiento del derecho, migración o administración de justicia?.

# ### 2. Excepciones Críticas y Contra-excepciones (Art. 6, apartado 3)

# Si el sistema cae en el **Anexo III**, puede quedar exento de la categoría de alto riesgo si no plantea un riesgo importante o no influye sustancialmente en la toma de decisiones. Sin embargo, hay un filtro infranqueable:

# *   **¿Realiza perfilado?**: ¿El sistema efectúa la elaboración de perfiles de personas físicas? (Si es "Sí", es **Alto Riesgo** automáticamente, sin posibilidad de excepción).
# *   **Condiciones de la Excepción (Art. 6.3, párrafo 2):** Si no realiza perfilado, ¿cumple alguna de estas condiciones?:
#     1.  Realiza una tarea procedimental limitada.
#     2.  Mejora el resultado de una actividad humana previa.
#     3.  Detecta patrones de decisión sin sustituir la valoración humana.
#     4.  Realiza solo una tarea preparatoria.

# ### 3. Variables y Lógica en Python

# Este bloque conecta con la salida del Artículo 5. Si el resultado anterior fue "CONTINUAR", ejecutamos esta lógica:

# ```python
def analizar_articulo_6():
    """
    Clasifica el sistema como Alto Riesgo o Riesgo Limitado/Mínimo según el Art. 6.
    """
    print("\n--- Clasificación de Riesgo (Artículo 6) ---")

    # --- Vía A: Art. 6(1) - Productos y Componentes de Seguridad ---
    es_producto_anexo_i = input("¿Es el sistema un producto o componente de seguridad del Anexo I (ej. máquinas, juguetes, dispositivos médicos)? (s/n): ").lower() == 's'
    evaluacion_terceros = input("¿La ley exige que este producto sea evaluado por un tercero (organismo notificado)? (s/n): ").lower() == 's'

    if es_producto_anexo_i and evaluacion_terceros:
        # Art. 6(1)(a) y (b)
        return "ALTO RIESGO: Sistema regulado por legislación de armonización del Anexo I."

    # --- Vía B: Art. 6(2) - Casos de uso del Anexo III ---
    en_anexo_iii = input("¿Se usa el sistema en biometría, educación, empleo, servicios públicos o policía (Anexo III)? (s/n): ").lower() == 's'

    if en_anexo_iii:
        # Art. 6(3) párrafo 3: El perfilado anula cualquier excepción
        realiza_perfilado = input("¿El sistema realiza perfilado de personas físicas (Art. 3, punto 52)? (s/n): ").lower() == 's'
        if realiza_perfilado:
             return "ALTO RIESGO: Los sistemas del Anexo III que realizan perfilado son siempre de alto riesgo (Art. 6(3))."

        # Art. 6(3) - Evaluación de influencia no sustancial
        print("Evaluando excepciones de riesgo no importante (Art. 6(3))...")
        tarea_limitada = input("¿Es una tarea procedimental muy limitada? (s/n): ").lower() == 's'
        mejora_humana = input("¿Solo mejora una actividad humana previa? (s/n): ").lower() == 's'
        preparatoria = input("¿Es solo una tarea preparatoria para una evaluación posterior? (s/n): ").lower() == 's'
        
        if tarea_limitada or mejora_humana or preparatoria:
            # Art. 6(4): Obligación de documentar la evaluación
            print("AVISO: El sistema NO es de alto riesgo, pero debe documentar esta evaluación antes de comercializarlo (Art. 6(4)).")
            return "RIESGO LIMITADO/MÍNIMO: Excepción por influencia no sustancial."
        else:
            return "ALTO RIESGO: Sistema incluido en el Anexo III sin excepciones aplicables."

    return "RIESGO LIMITADO/MÍNIMO: El sistema no entra en las categorías de alto riesgo."
# ```

# ### 4. Destino del Flujo

# *   **Si es ALTO RIESGO:** El usuario debe ser redirigido a las **Obligaciones para Sistemas de Alto Riesgo (Artículos 8 a 15)** (gestión de riesgos, datos, documentación, transparencia técnica, supervisión humana, etc.).
# *   **Si es RIESGO LIMITADO/MÍNIMO:** El flujo continúa hacia las **Obligaciones de Transparencia (Artículo 50)** (ej. informar que se interactúa con una IA, marcar deepfakes, etc.).

# ¿Deseas que continuemos analizando las **obligaciones técnicas para sistemas de Alto Riesgo (Sección 2, Artículos 8 al 15)** o prefieres saltar a los **Modelos de IA de Uso General (GPAI)**?

resultado = analizar_articulo_6()
print(f"\nResultado de la evaluación: {resultado}")
````
